### Data Preprocessing & Splitting for English-to-Mwaghavul NMT
We will inspect the parallel text columns, handle any missing values, clean up whitespace, and perform a stratified or standard split into Training (80%), Validation (10%), and Testing (10%) sets.

```python
from sklearn.model_selection import train_test_split
import pandas as pd

# We will use df_xlsx (or df_csv) as our master translation dataframe
df_clean = df_xlsx[['English', 'Mwaghavul']].dropna().copy()

# Strip whitespaces
df_clean['English'] = df_clean['English'].astype(str).str.strip()
df_clean['Mwaghavul'] = df_clean['Mwaghavul'].astype(str).str.strip()

# Filter out any empty rows resulting from stripping
df_clean = df_clean[(df_clean['English'] != '') & (df_clean['Mwaghavul'] != '')]

print(f"Total parallel sentences available after cleaning: {len(df_clean)}")

# Split into Train (80%) and Temp (20%)
train_df, temp_df = train_test_split(df_clean, test_size=0.20, random_state=42)

# Split Temp into Validation (10% total) and Test (10% total)
val_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=42)

print(f"Train set size: {len(train_df)}")
print(f"Validation set size: {len(val_df)}")
print(f"Test set size: {len(test_df)}")

# Display a few samples
display(train_df.head(5))
```

### Dataset Tokenization and Model Setup
We will convert the dataframes into Hugging Face datasets and tokenize the English and Mwaghavul parallel corpora using the `google/mt5-small` model's tokenizer.

```python
from datasets import Dataset, DatasetDict
from transformers import AutoTokenizer

# 1. Convert Pandas DataFrames to Hugging Face Datasets
dataset_dict = DatasetDict({
    'train': Dataset.from_pandas(train_df.reset_index(drop=True)),
    'validation': Dataset.from_pandas(val_df.reset_index(drop=True)),
    'test': Dataset.from_pandas(test_df.reset_index(drop=True))
})

# 2. Load the mT5 tokenizer
model_checkpoint = "google/mt5-small"
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)

# Prefix for translation task (recommended for T5 style models)
prefix = "translate English to Mwaghavul: "

def preprocess_function(examples):
    inputs = [prefix + str(ex) for ex in examples['English']]
    targets = [str(ex) for ex in examples['Mwaghavul']]

    model_inputs = tokenizer(inputs, max_length=128, truncation=True)

    # Setup the detector/labels for targets
    labels = tokenizer(text_target=targets, max_length=128, truncation=True)
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# 3. Tokenize dataset
tokenized_datasets = dataset_dict.map(preprocess_function, batched=True)
print("Tokenization completed successfully!")
print(tokenized_datasets)
```

### Initialize Model and Configure training
We will import the model class, define training configurations, load the validation metrics, and set up our Seq2SeqTrainer.

```python
import numpy as np
import evaluate
from transformers import AutoModelForSeq2SeqLM, DataCollatorForSeq2Seq, Seq2SeqTrainingArguments, Seq2SeqTrainer

# 1. Load the model
model = AutoModelForSeq2SeqLM.from_pretrained(model_checkpoint)

# 2. Data Collator for padding
data_collator = DataCollatorForSeq2Seq(tokenizer, model=model)

# 3. Load BLEU metric
metric = evaluate.load("sacrebleu")

def postprocess_text(preds, labels):
    preds = [pred.strip() for pred in preds]
    labels = [[label.strip()] for label in labels]
    return preds, labels

def compute_metrics(eval_preds):
    preds, labels = eval_preds
    if isinstance(preds, tuple):
        preds = preds[0]

    decoded_preds = tokenizer.batch_decode(preds, skip_special_tokens=True)

    # Replace -100 in labels as we cannot decode them
    labels = np.where(labels != -100, labels, tokenizer.pad_token_id)
    decoded_labels = tokenizer.batch_decode(labels, skip_special_tokens=True)

    decoded_preds, decoded_labels = postprocess_text(decoded_preds, decoded_labels)

    result = metric.compute(predictions=decoded_preds, references=decoded_labels)
    return {"bleu": result["score"]}

# 4. Define training configurations
training_args = Seq2SeqTrainingArguments(
    output_dir="./results_nmt",
    evaluation_strategy="epoch",
    learning_rate=3e-4,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    weight_decay=0.01,
    save_total_limit=2,
    num_train_epochs=3,
    predict_with_generate=True,
    fp16=True, # Recommended on GPU
    logging_steps=100,
    report_to="none"
)

# 5. Build Trainer
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["validation"],
    tokenizer=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

print("Model and Trainer are configured successfully!")
```

### Train the Neural Machine Translation Model
We will execute `.train()` to fine-tune the `mT5-small` translator on our parallel dataset, followed by evaluating the validation BLEU score.

```python
import numpy as np
import evaluate
import torch
from datasets import Dataset, DatasetDict
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, DataCollatorForSeq2Seq, Seq2SeqTrainingArguments, Seq2SeqTrainer

# 1. Prepare HF Datasets from our splits
dataset_dict = DatasetDict({
    'train': Dataset.from_pandas(train_df.reset_index(drop=True)),
    'validation': Dataset.from_pandas(val_df.reset_index(drop=True)),
    'test': Dataset.from_pandas(test_df.reset_index(drop=True))
})

# 2. Setup Tokenizer & Model
model_checkpoint = "google/mt5-small"
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)
model = AutoModelForSeq2SeqLM.from_pretrained(model_checkpoint)

prefix = "translate English to Mwaghavul: "

def preprocess_function(examples):
    inputs = [prefix + str(ex) for ex in examples['English']]
    targets = [str(ex) for ex in examples['Mwaghavul']]
    model_inputs = tokenizer(inputs, max_length=128, truncation=True)
    labels = tokenizer(text_target=targets, max_length=128, truncation=True)
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

tokenized_datasets = dataset_dict.map(preprocess_function, batched=True)

# 3. Setup Metric & Collator
data_collator = DataCollatorForSeq2Seq(tokenizer, model=model)
metric = evaluate.load("sacrebleu")

def postprocess_text(preds, labels):
    preds = [pred.strip() for pred in preds]
    labels = [[label.strip()] for label in labels]
    return preds, labels

def compute_metrics(eval_preds):
    preds, labels = eval_preds
    if isinstance(preds, tuple):
        preds = preds[0]
    decoded_preds = tokenizer.batch_decode(preds, skip_special_tokens=True)
    labels = np.where(labels != -100, labels, tokenizer.pad_token_id)
    decoded_labels = tokenizer.batch_decode(labels, skip_special_tokens=True)
    decoded_preds, decoded_labels = postprocess_text(decoded_preds, decoded_labels)
    result = metric.compute(predictions=decoded_preds, references=decoded_labels)
    return {"bleu": result["score"]}

# 4. Configure Training Arguments
training_args = Seq2SeqTrainingArguments(
    output_dir="./results_nmt",
    evaluation_strategy="epoch",
    learning_rate=3e-4,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    weight_decay=0.01,
    save_total_limit=2,
    num_train_epochs=3,
    predict_with_generate=True,
    fp16=torch.cuda.is_available(), # Use mixed precision if a GPU is available
    logging_steps=100,
    report_to="none"
)

# 5. Initialize Trainer
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["validation"],
    tokenizer=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

# 6. Run Training & Evaluation
print("Starting Model Training...")
trainer.train()

eval_results = trainer.evaluate()
print("\n--- Evaluation Results on Validation Set ---")
print(eval_results)
```

### Test the Translation Model (Inference)
Let's write a function to translate arbitrary English sentences using our fine-tuned model and check some translation examples.

```python
import re
import torch

def translate_sentence(english_sentence, model, tokenizer, device="cuda"):
    input_text = "translate English to Mwaghavul: " + english_sentence
    inputs = tokenizer(input_text, return_tensors="pt", max_length=128, truncation=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    # T5 models expect the decoder to start with <pad> followed by <extra_id_0> to trigger text generation
    decoder_start_ids = tokenizer(["<pad> <extra_id_0>"], return_tensors="pt", add_special_tokens=False)["input_ids"].to(device)

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            decoder_input_ids=decoder_start_ids,
            max_new_tokens=128,
            num_beams=1,
            do_sample=False
        )

    # Decode the entire output sequence including special tokens
    translation = tokenizer.decode(output_ids[0], skip_special_tokens=False)

    # Clean up the pretraining sequence and helper markers to retrieve only the raw Mwaghavul text
    translation = re.sub(r'<pad>|<extra_id_\d+>|</s>', '', translation).strip()
    return translation

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

test_sentences = [
    "You should come with us on time.",  # From training dataset
    "Put the book on the bed.",
    "The room is so warm."
]

print("--- Translation Results ---")
for sentence in test_sentences:
    translation = translate_sentence(sentence, model, tokenizer, device=device)
    print(f"English: {sentence}")
    print(f"Mwaghavul Translation: {translation}\n")
```
